```python
import csv
import os
import shutil
from datetime import datetime
from glob import glob

class SimpleLogger:
    def info(self, msg): print("[INFO]", msg)
    def warning(self, msg): print("[WARN]", msg)
    def error(self, msg): print("[ERROR]", msg)

logger = SimpleLogger()


def main():
    #HARDCODE PATH DI SINI
    # input_dir   = "output/broker_summary"
    input_dir   = "output/broks"
    archive_dir = "archive/trash/march"
    output_dir  = "data/broker_summary/2026"

    # cari semua file CSV
    pattern = os.path.join(input_dir, "*.csv")
    csv_files = sorted(glob(pattern))

    if not csv_files:
        logger.error(f"Tidak ada file CSV di '{input_dir}'. Pastikan path sudah benar.")
        return

    logger.info(f"Ditemukan {len(csv_files)} file CSV di '{input_dir}'")

    # load all filea
    all_rows   = []
    fieldnames = None
    skipped    = []

    for fpath in csv_files:
        try:
            with open(fpath, "r", encoding="utf-8", newline="") as f:
                reader = csv.DictReader(f)

                if fieldnames is None:
                    fieldnames = reader.fieldnames
                    if not fieldnames:
                        logger.warning(f"Header kosong: {fpath}, skip.")
                        skipped.append(fpath)
                        continue

                row_count = 0
                for row in reader:
                    all_rows.append(row)
                    row_count += 1

            logger.info(f"Read {row_count} rows ← {os.path.basename(fpath)}")

        except Exception as e:
            logger.error(f"Gagal baca {fpath}: {e}")
            skipped.append(fpath)

    if not all_rows:
        logger.error("Tidak ada data yang berhasil dibaca.")
        return

    #  merge
    os.makedirs(output_dir, exist_ok=True)
    ts = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
    out_path = os.path.join(output_dir, f"broksum_march.csv")

    with open(out_path, "w", encoding="utf-8", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames, extrasaction="ignore")
        writer.writeheader()
        writer.writerows(all_rows)

    logger.info(f"Merged {len(all_rows)} rows -> {out_path}")

    # archive file
    os.makedirs(archive_dir, exist_ok=True)

    for fpath in csv_files:
        if fpath in skipped:
            continue
        try:
            dest = os.path.join(archive_dir, os.path.basename(fpath))
            if os.path.exists(dest):
                base, ext = os.path.splitext(os.path.basename(fpath))
                dest = os.path.join(archive_dir, f"{base}_{ts}{ext}")
            shutil.move(fpath, dest)
        except Exception as e:
            logger.error(f"Gagal pindah {fpath}: {e}")

    print("SELESAI")
    print("Output:", out_path)

main()
```
