```python
import json
import os
from datetime import datetime
from glob import glob
```

```python
# HELPER FUNCTIONS
def load_json_file(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    if not isinstance(data, list):
        raise ValueError(f"File {file_path} tidak berisi list JSON")
    return data

def safe_date(item):
    """Return datetime object for sorting/filtering"""
    date_str = item.get("date")
    if not date_str:
        return datetime(1900, 1, 1)
    try:
        return datetime.fromisoformat(date_str[:10])
    except (ValueError, TypeError):
        return datetime(1900, 1, 1)

def filter_fields(item):
    """Keep only required fields"""
    return {
        "title": item.get("title", ""),
        "date": item.get("date", ""),
        "source": item.get("source", ""),
        "full_content": item.get("full_content", ""),
        "summary": item.get("summary", "")
    }

def load_all_json_from_folder(folder_path):
    all_data = []
    json_files = glob(os.path.join(folder_path, "*.json"))
    for fpath in json_files:
        data = load_json_file(fpath)
        all_data.extend(data)
    return all_data
```

# `khusus 2026`

```python
idxchannel_detail = "/Users/albert/Documents/Finances/data/raw/alternative_data/news/idx_channel/detailed"
kabarbursa_detail = "/Users/albert/Documents/Finances/data/raw/alternative_data/news/kabar_bursa/kabarbursa_detailed"
```

```python
# OUTPUT PATH
goal_path = "/Users/albert/Documents/Finances/data/processed/market"
os.makedirs(goal_path, exist_ok=True)
output_file = os.path.join(goal_path, "market_news_2026.json")

# HELPER FUNCTIONS
def load_json_file(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    if not isinstance(data, list):
        raise ValueError(f"File {file_path} tidak berisi list JSON")
    return data

def safe_date(item):
    """Return datetime object for sorting/filtering"""
    date_str = item.get("date")
    if not date_str:
        return datetime(1900, 1, 1)
    try:
        return datetime.fromisoformat(date_str[:10])
    except (ValueError, TypeError):
        return datetime(1900, 1, 1)

def filter_fields(item):
    """Keep only required fields"""
    return {
        "title": item.get("title", ""),
        "date": item.get("date", ""),
        "source": item.get("source", ""),
        "full_content": item.get("full_content", ""),
        "summary": item.get("summary", "")
    }

def load_all_json_from_folder(folder_path):
    all_data = []
    json_files = glob(os.path.join(folder_path, "*.json"))
    for fpath in json_files:
        data = load_json_file(fpath)
        all_data.extend(data)
    return all_data

# LOAD DATA
data_idx = load_all_json_from_folder(idxchannel_detail)
data_kb = load_all_json_from_folder(kabarbursa_detail)

print(f"📄 IDX Channel articles: {len(data_idx)}")
print(f"📄 Kabar Bursa articles: {len(data_kb)}")

# MERGE
all_data = data_idx + data_kb
print(f"📊 Total sebelum sort: {len(all_data)}")

# FILTER 2026 KE ATAS
start_date = datetime(2026, 1, 1)
filtered_data = [item for item in all_data if safe_date(item) >= start_date]

print(f"📊 Total setelah filter 2026 ke atas: {len(filtered_data)}")

# SORT BY DATE
filtered_data.sort(key=safe_date)

# KEEP ONLY REQUIRED FIELDS
final_data = [filter_fields(item) for item in filtered_data]

# SAVE OUTPUT
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(final_data, f, indent=2, ensure_ascii=False)

print("✅ Merge & filter selesai")
print(f"📁 File hasil: {output_file}")
print(f"📊 Total final: {len(final_data)}")
```
