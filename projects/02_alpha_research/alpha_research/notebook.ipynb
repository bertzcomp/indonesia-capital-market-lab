```python
import polars as pl

df = pl.read_parquet("/Users/albert/Documents/Finances/projects/02_alpha_research/alpha_research/data/features/live/latest/base_features.parquet")

recent = (
    df.filter(pl.col("date") >= pl.date(2026, 5, 19))
    .group_by("date")
    .agg([
        pl.len().alias("initial"),
        pl.col("close").is_not_null().sum().alias("valid_close"),
        ((pl.col("close") >= 100) & (pl.col("close") <= 1000)).sum().alias("price_100_1000"),
        (pl.col("traded_value_proxy").fill_null(0) >= 500_000_000).sum().alias("min_traded_value"),
        (pl.col("has_broksum").fill_null(0) == 1).sum().alias("has_broksum"),
        (
            (pl.col("close").is_not_null()) &
            (pl.col("close") >= 100) &
            (pl.col("close") <= 1000) &
            (pl.col("traded_value_proxy").fill_null(0) >= 500_000_000) &
            (pl.col("has_broksum").fill_null(0) == 1)
        ).sum().alias("final_eligible"),
    ])
    .sort("date")
)

print(recent)
```

# BI RATE Scraper

```python
# CONFIG
api_key = "cdc65504d0b176f0b763f2d8dcd855dd"

import requests
import time
import json
import os
import random

api_key = ""

years = {
    2026: 126,
    2025: 125,
    2024: 124,
    2023: 123,
    2022: 122,
    2021: 121,
    2020: 120,
    2019: 119,
    2018: 118,
    2017: 117,
    2016: 116,
    2015: 115,
}

base_url = "https://webapi.bps.go.id/v1/api/list/model/data/lang/eng/domain/0000/var/379/th/{th}/key/{key}"

output_dir = "/Users/albert/Documents/Finances/projects/02_alpha_research/alpha_research/data/raw/bps_bi_rate"
os.makedirs(output_dir, exist_ok=True)

log_file = os.path.join(output_dir, "download_log.txt")

# FUNCTION: FETCH WITH RETRY
def fetch_bps(th, api_key, max_retry=3):
    url = base_url.format(th=th, key=api_key)
    for attempt in range(1, max_retry + 1):
        try:
            r = requests.get(url, timeout=30)
            # sukses
            if r.status_code == 200:
                return r.json()
            print(f"HTTP {r.status_code} (attempt {attempt})")
        except Exception as e:
            print(f"Exception (attempt {attempt}): {e}")
        # exponential backoff
        sleep_time = (2 ** attempt) + random.random()
        print(f"RETRY in {sleep_time:.2f}s")
        time.sleep(sleep_time)
    raise Exception(f"Failed after {max_retry} retries for th={th}")

# MAIN LOOP
for year, th in years.items():
    file_path = os.path.join(output_dir, f"bi_rate_{year}.json")
    # RESUME LOGIC (skip jika sudah ada)
    if os.path.exists(file_path):
        print(f"SKIP {year} (already exists)")
        continue
    print(f"\n FETCHING {year}...")
    try:
        data = fetch_bps(th, api_key)
        # validasi status API
        if data.get("status") != "OK":
            print(f"!!! {year} status NOT OK")
            continue
        # simpan file
        with open(file_path, "w", encoding="utf-8") as f:
            json.dump(data, f, indent=2, ensure_ascii=False)
        print(f"SAVED {file_path}")
        # log success
        with open(log_file, "a") as log:
            log.write(f"SUCCESS {year}\n")
    except Exception as e:
        print(f"FAILED {year}: {e}")
        # log failure
        with open(log_file, "a") as log:
            log.write(f"FAILED {year}: {e}\n")

    # SAFE DELAY (anti overload)
    sleep_time = 1.5 + random.random()
    print(f"SLEEP {sleep_time:.2f}s\n")
    time.sleep(sleep_time)

print("\nDONE - All available data processed")
```

```python
import polars as pl

paths = [
    "data/features/history/base_features.parquet",
    "data/features/live/latest/base_features.parquet",
]

for p in paths:
    df = pl.read_parquet(p)
    numeric_cols = [
        c for c, d in zip(df.columns, df.dtypes)
        if d in (pl.Float32, pl.Float64)
    ]

    df = df.with_columns([
        pl.when(pl.col(c).is_nan() | pl.col(c).is_infinite())
          .then(None)
          .otherwise(pl.col(c))
          .alias(c)
        for c in numeric_cols
    ])

    df.write_parquet(p)
    print("cleaned", p, df.shape)
```

## Live feature view dan label view

```python
import polars as pl
from pathlib import Path

base = Path("data/news/event_intelligence")
df = pl.read_parquet(base / "news_event_intelligence_dataset.parquet")

id_cols = [
    "event_row_id", "article_id", "news_date", "entry_date", "ticker",
]

metadata_cols = [
    "source", "news_category",
    "sector", "subsector", "industry", "subindustry", "listing_board",
]

event_cols = [
    "event_scope", "event_type", "event_side", "impact_channel",
    "materiality_label", "materiality_score",
    "news_intensity_score", "uncertainty_score", "novelty_score",
    "bullish_keyword_hits", "bearish_keyword_hits", "uncertainty_keyword_hits",
    "same_event_seen_so_far",
]

live_market_cols = [
    "open", "high", "low", "close", "volume", "value", "frequency",
    "foreign_buy", "foreign_sell", "daily_ret", "is_zero_volume",
    "avg_volume_20d", "volume_ratio", "bwd_volatility_20d", "drawdown_20d",
]

bwd_cols = [
    c for c in df.columns
    if c.startswith("bwd_ret_") or c.startswith("bwd_volume_ratio_")
]

live_cols = [
    c for c in id_cols + metadata_cols + event_cols + live_market_cols + bwd_cols
    if c in df.columns
]

label_cols = [
    c for c in df.columns
    if (
        c.startswith("fwd_ret_")
        or c.startswith("ihsg_fwd_ret_")
        or c.startswith("sector_fwd_ret_")
        or c.startswith("market_alpha_")
        or c.startswith("sector_alpha_")
        or c.startswith("volatility_shock_")
        or c.startswith("mae_")
        or c.startswith("mfe_")
        or c.endswith("_label_5d")
        or c == "reaction_label_5d"
    )
]

label_view_cols = [c for c in id_cols + label_cols if c in df.columns]

live = df.select(live_cols)
labels = df.select(label_view_cols)

live.write_parquet(base / "news_event_live_features.parquet")
labels.write_parquet(base / "news_event_training_labels.parquet")

live.head(10).write_csv(base / "sample_live_features_10.csv")
labels.head(10).write_csv(base / "sample_training_labels_10.csv")

print("Full dataset:", df.shape)
print("Live features:", live.shape)
print("Training labels:", labels.shape)
```

# Event study detail

```python
import polars as pl
from pathlib import Path

df = pl.read_parquet("data/news/event_intelligence/news_event_intelligence_dataset.parquet")
out = Path("data/news/event_intelligence/report")
out.mkdir(parents=True, exist_ok=True)

def event_study(group_cols, filename):
    report = (
        df.group_by(group_cols)
          .agg([
              pl.len().alias("n_rows"),
              pl.col("article_id").n_unique().alias("n_articles"),
              pl.col("ticker").n_unique().alias("n_tickers"),

              pl.col("sector_alpha_5d").mean().alias("avg_sector_alpha_5d"),
              pl.col("sector_alpha_5d").median().alias("median_sector_alpha_5d"),
              (pl.col("sector_alpha_5d") > 0).mean().alias("hit_rate_sector_alpha_5d"),
              (pl.col("sector_alpha_5d") > 0.01).mean().alias("hit_rate_sector_alpha_gt_1pct_5d"),

              pl.col("market_alpha_5d").mean().alias("avg_market_alpha_5d"),
              (pl.col("market_alpha_5d") > 0).mean().alias("hit_rate_market_alpha_5d"),

              pl.col("volatility_shock_5d").mean().alias("avg_volatility_shock_5d"),
              (pl.col("volatility_shock_5d") > 1.5).mean().alias("volatility_shock_rate_5d"),

              pl.col("mae_5d").mean().alias("avg_mae_5d"),
              pl.col("mfe_5d").mean().alias("avg_mfe_5d"),

              (pl.col("sentiment_trap_label_5d") == 1).mean().alias("sentiment_trap_rate_5d"),
              (pl.col("delayed_reaction_label_5d") == 1).mean().alias("delayed_reaction_rate_5d"),
              (pl.col("acceleration_trigger_label_5d") == 1).mean().alias("acceleration_trigger_rate_5d"),
          ])
          .filter(pl.col("n_rows") >= 30)
          .sort("n_rows", descending=True)
    )
    report.write_csv(out / filename)
    return report

event_study(["event_type"], "event_study_by_event_type_detailed.csv")
event_study(["event_type", "event_side"], "event_study_by_event_type_side.csv")
event_study(["event_type", "sector"], "event_study_by_event_type_sector.csv")
event_study(["event_type", "news_category"], "event_study_by_event_type_category.csv")

print("saved to", out)
```
