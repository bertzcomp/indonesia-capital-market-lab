```python
import json
import os


def merge_market_news(folder_path, output_path):
    
    def load_json(path): 
        with open(path, "r", encoding="utf-8") as f:
            return json.load(f)
    
    def normalize_idx(data):
        return [{
            "title": item["title"].strip(),
            "date": item["date"],
            "source": item.get("source", "IDX Channel"),
            "full_content": item.get("full_content", ""),
            "summary": item.get("summary", "")
        } for item in data]

    def normalize_kabar(data):
        return [{
            "title": item["title"].strip(),
            "date": item["date"],
            "source": "Kabar Bursa",
            "full_content": item.get("full_content", ""),
            "summary": item.get("summary", "")
        } for item in data]

    def deduplicate_title(news):
        seen = set()
        cleaned = []
        for item in news:
            title = item.get("title", "").lower().strip()
            if title not in seen:
                seen.add(title)
                cleaned.append(item)
        print("After title dedup:", len(cleaned))
        return cleaned

    print("Scanning folder...")
    
    combined = [] 
    for file in os.listdir(folder_path):
        if not file.endswith(".json"):
            continue
        path = os.path.join(folder_path, file)
        
        print("Loadng:", file)
        
        data = load_json(path)
        filename = file.lower()
        # pilih normalizer
        if "idx" in filename:
            print("Applying IDX normalization")
            data = normalize_idx(data)
        elif "kabar" in filename:
            print("Applying Kabar normalization")
            data = normalize_kabar(data)
        # selain itu dianggap market/news biasa
        combined.extend(data)

    print("Total before dedup:", len(combined))

    combined = deduplicate_title(combined)

    print("Sorting by date...")

    combined.sort(key=lambda x: x.get("date", ""))

    print("Saving...")

    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(combined, f, indent=2, ensure_ascii=False)

    print("Merge selesai.")
    print("Total news:", len(combined))

    return combined
```
