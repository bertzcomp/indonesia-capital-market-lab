```python
import requests
import xml.etree.ElementTree as ET
from datetime import datetime, timedelta

# Konfigurasi API key — isi di sini atau gunakan Colab Secrets
FRED_API_KEY = ""   # <- stlouisfed.org API key
BPS_API_KEY  = ""   # <- BPS API key

_today      = datetime.today()
_year_start = f"{_today.year}-01-01"
_yesterday  = (_today - timedelta(days=1)).strftime("%Y-%m-%d")

# ── 1. USD/IDR dari Bank Indonesia ──────────────────────────────────
def fetch_usdidr_bi(start, end):
    url = (
        "https://www.bi.go.id/biwebservice/wskursbi.asmx/getSubKursLokal3"
        f"?mts=USD&startdate={start}&enddate={end}"
    )

    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    for i in range(3):  # retry 3x
        try:
            resp = requests.get(url, headers=headers, timeout=30)
            resp.raise_for_status()
            break
        except Exception as e:
            print(f"Retry {i+1} BI error: {e}")
            if i == 2:
                return pd.DataFrame()

    root = ET.fromstring(resp.content)
    rows = []
    for table in root.iter("Table"):
        tgl_el  = table.find("tgl_subkurslokal")
        beli_el = table.find("beli_subkurslokal")
        jual_el = table.find("jual_subkurslokal")
        if tgl_el is None or beli_el is None or jual_el is None:
            continue
        try:
            tgl = pd.to_datetime(tgl_el.text).strftime("%Y-%m-%d")
            mid = (float(beli_el.text) + float(jual_el.text)) / 2
            rows.append({"date": tgl, "usdidr": mid})
        except Exception:
            continue

    if not rows:
        print("  \u26a0\ufe0f  BI USD/IDR: tidak ada data dari XML")
        return pd.DataFrame()

    return (pd.DataFrame(rows)
              .drop_duplicates("date")
              .sort_values("date")
              .reset_index(drop=True))


# ── 2. FRED series (brent daily & coal monthly) ───────────────────────
def fetch_fred(series_id, col_rename, start, end, api_key):
    url = "https://api.stlouisfed.org/fred/series/observations"
    params = {
        "series_id": series_id,
        "api_key": api_key,
        "file_type": "json",
        "observation_start": start,
        "observation_end": end,
    }
    try:
        resp = requests.get(url, params=params, timeout=30)
        resp.raise_for_status()
        data = resp.json()
    except Exception as e:
        print(f"  \u26a0\ufe0f  FRED {series_id} error: {e}")
        return pd.DataFrame()

    if "observations" not in data:
        print(f"  \u26a0\ufe0f  FRED {series_id}: {data.get('error_message', 'no observations key')}")
        return pd.DataFrame()

    rows = [
        {"date": o["date"], col_rename: float(o["value"])}
        for o in data["observations"] if o["value"] != "."
    ]
    if not rows:
        return pd.DataFrame()

    return (pd.DataFrame(rows)
              .drop_duplicates("date")
              .sort_values("date")
              .reset_index(drop=True))


# ── 3. BI Rate dari BPS ─────────────────────────────────────────────
def fetch_bi_rate_bps(api_key):
    import requests
    import pandas as pd
    from datetime import datetime

    today = datetime.today()
    year = today.year
    year_bps = year - 1900  # format BPS

    url = (
        f"https://webapi.bps.go.id/v1/api/list/model/data/lang/ind"
        f"/domain/0000/var/379/th/{year_bps}/key/{api_key}"
    )

    try:
        resp = requests.get(url, timeout=30)
        resp.raise_for_status()
        data = resp.json()
    except Exception as e:
        print(f"⚠️ BPS BI Rate fetch error: {e}")
        return pd.DataFrame()

    # Validasi response
    if data.get("status") != "OK":
        print(f"⚠️ BPS status not OK: {data.get('status')}")
        return pd.DataFrame()

    content = data.get("datacontent", {})
    if not content:
        print("⚠️ BPS: datacontent kosong")
        return pd.DataFrame()

    rows = []

    for k, v in content.items():
        try:
            # Extract bulan dari key (robust)
            # handle bulan 1 digit & 2 digit
            last2 = int(k[-2:])
            if 1 <= last2 <= 12:
                month = last2
            else:
                month = int(k[-1])

            # skip "Tahunan" (biasanya 13)
            if month > 12:
                continue

            rows.append({
                "date": f"{year}-{month:02d}-01",
                "bi_rate_pct": float(v)
            })

        except Exception:
            continue

    if not rows:
        print("⚠️ BPS: tidak ada data valid setelah parsing")
        return pd.DataFrame()

    df = (
        pd.DataFrame(rows)
        .drop_duplicates("date")
        .sort_values("date")
        .reset_index(drop=True)
    )

    print(f"✅ BI Rate BPS: {len(df)} rows ({df['date'].min()} → {df['date'].max()})")

    return df


# ── Fetch semua ───────────────────────────────────────────────────────
print("Fetching macro data dari API...")

def _fallback_from_master(col):
    if col in master.columns:
        return (master[["date", col]]
                .dropna(subset=[col])
                .drop_duplicates("date")
                .sort_values("date")
                .reset_index(drop=True))
    return pd.DataFrame()

df_usdidr = fetch_usdidr_bi(_year_start, _yesterday)
if df_usdidr.empty:
    print("  \u26a0\ufe0f  usdidr   : fallback dari master")
    df_usdidr = _fallback_from_master("usdidr")
else:
    print(f"  \u2705 usdidr   : {len(df_usdidr)} rows "
          f"({df_usdidr['date'].min()} \u2192 {df_usdidr['date'].max()})")

df_brent = fetch_fred("DCOILBRENTEU", "brent_usd", _year_start, _yesterday, FRED_API_KEY)
if df_brent.empty:
    print("  \u26a0\ufe0f  brent_usd: fallback dari master")
    df_brent = _fallback_from_master("brent_usd")
else:
    print(f"  \u2705 brent_usd: {len(df_brent)} rows "
          f"({df_brent['date'].min()} \u2192 {df_brent['date'].max()})")

df_coal_m = fetch_fred("PCOALAUUSDM", "coal_usd", _year_start, _yesterday, FRED_API_KEY)
if df_coal_m.empty:
    print("  \u26a0\ufe0f  coal_usd : fallback dari master (monthly)")
    df_coal_m = _fallback_from_master("coal_usd")
else:
    print(f"  \u2705 coal_usd : {len(df_coal_m)} rows bulanan "
          f"({df_coal_m['date'].min()} \u2192 {df_coal_m['date'].max()})")

df_bi_rate = fetch_bi_rate_bps(BPS_API_KEY)
if df_bi_rate.empty:
    print("  \u26a0\ufe0f  bi_rate  : fallback dari master")
    df_bi_rate = _fallback_from_master("bi_rate_pct")
else:
    print(f"  \u2705 bi_rate  : {len(df_bi_rate)} rows "
          f"({df_bi_rate['date'].min()} \u2192 {df_bi_rate['date'].max()})")

# ── Simpan CSV macro ke MACRO_DIR ────────────────────────────────────
Path(MACRO_DIR).mkdir(parents=True, exist_ok=True)
_csv_map = [
    (df_usdidr,  "usdidr.csv",   "usdidr"),
    (df_brent,   "brent.csv",    "brent_usd"),
    (df_coal_m,  "coal.csv",     "coal_usd"),
    (df_bi_rate, "bi_rate.csv",  "bi_rate_pct"),
]
for _df, _fname, _col in _csv_map:
    if not _df.empty and _col in _df.columns:
        _df[["date", _col]].to_csv(f"{MACRO_DIR}/{_fname}", index=False)

# ── Buat macro_ref per tanggal baru ──────────────────────────────────
new_dates_sorted = sorted(new_merged["date"].unique())
macro_ref = pd.DataFrame({"date": new_dates_sorted})

def _ffill_merge_daily(ref_df, src_df, col):
    """Left-join harian lalu forward-fill (untuk data harian/sparse)."""
    if src_df.empty or col not in src_df.columns:
        return ref_df
    ref_df = ref_df.merge(src_df[["date", col]], on="date", how="left")
    ref_df[col] = ref_df[col].ffill().bfill()
    return ref_df

def _ffill_monthly_to_daily(ref_df, src_df, col):
    """Reindex series bulanan ke indeks harian lalu forward-fill."""
    if src_df.empty or col not in src_df.columns:
        return ref_df
    _date_idx = pd.to_datetime(ref_df["date"])
    _s = (src_df
          .assign(date=lambda d: pd.to_datetime(d["date"]))
          .set_index("date")[col]
          .reindex(_date_idx)
          .ffill().bfill())
    ref_df[col] = _s.values
    return ref_df

# Brent: harian — left-join + ffill (bisa ada gap weekend/libur)
macro_ref = _ffill_merge_daily(macro_ref, df_brent,  "brent_usd")

# USDIDR: harian — left-join + ffill (BI tidak publish tiap hari)
macro_ref = _ffill_merge_daily(macro_ref, df_usdidr, "usdidr")

# Coal: bulanan → forward-fill ke harian
macro_ref = _ffill_monthly_to_daily(macro_ref, df_coal_m, "coal_usd")

# BI Rate: bulanan → forward-fill ke harian
macro_ref = _ffill_monthly_to_daily(macro_ref, df_bi_rate, "bi_rate_pct")

# Merge macro ke new_merged
_macro_cols = [c for c in ["usdidr","brent_usd","coal_usd","bi_rate_pct"]
               if c in macro_ref.columns]
new_merged = new_merged.merge(macro_ref[["date"] + _macro_cols], on="date", how="left")

print("\nNull check macro pada baris baru:")
for c in _macro_cols:
    null_pct = new_merged[c].isna().mean() * 100
    icon = "\u2705" if null_pct == 0 else "\u26a0\ufe0f "
    print(f"  {icon} {c:<12}: {null_pct:.1f}% null")

# STEP 7: Merge Sektor dari master
sektor_cols = [c for c in ["ticker","Sektor","SubSektor","Industri","SubIndustri"]
               if c in master.columns]
if len(sektor_cols) > 1:
    sektor_ref = master[sektor_cols].drop_duplicates("ticker").reset_index(drop=True)
    new_merged = new_merged.merge(sektor_ref, on="ticker", how="left")
    print(f"Sektor merged: {new_merged['Sektor'].notna().sum():,} rows punya sektor")

print(f"\nFinal new_merged: {new_merged.shape}")
```

```python
import requests
import xml.etree.ElementTree as ET
from datetime import datetime, timedelta

# Konfigurasi API key — isi di sini atau gunakan Colab Secrets
FRED_API_KEY = "4409799e0f6ca420dc77637f7a65d71e"   # <- stlouisfed.org API key
BPS_API_KEY  = "cdc65504d0b176f0b763f2d8dcd855dd"   # <- BPS API key

_today      = datetime.today()
_year_start = f"{_today.year}-01-01"
_yesterday  = (_today - timedelta(days=1)).strftime("%Y-%m-%d")

# ── 1. USD/IDR dari Bank Indonesia ──────────────────────────────────
def fetch_usdidr_bi(start, end):
    url = (
        "https://www.bi.go.id/biwebservice/wskursbi.asmx/getSubKursLokal3"
        f"?mts=USD&startdate={start}&enddate={end}"
    )

    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    for i in range(3):  # retry 3x
        try:
            resp = requests.get(url, headers=headers, timeout=30)
            resp.raise_for_status()
            break
        except Exception as e:
            print(f"Retry {i+1} BI error: {e}")
            if i == 2:
                return pd.DataFrame()

    root = ET.fromstring(resp.content)
    rows = []
    for table in root.iter("Table"):
        tgl_el  = table.find("tgl_subkurslokal")
        beli_el = table.find("beli_subkurslokal")
        jual_el = table.find("jual_subkurslokal")
        if tgl_el is None or beli_el is None or jual_el is None:
            continue
        try:
            tgl = pd.to_datetime(tgl_el.text).strftime("%Y-%m-%d")
            mid = (float(beli_el.text) + float(jual_el.text)) / 2
            rows.append({"date": tgl, "usdidr": mid})
        except Exception:
            continue

    if not rows:
        print("  \u26a0\ufe0f  BI USD/IDR: tidak ada data dari XML")
        return pd.DataFrame()

    return (pd.DataFrame(rows)
              .drop_duplicates("date")
              .sort_values("date")
              .reset_index(drop=True))


# ── 2. FRED series (brent daily & coal monthly) ───────────────────────
def fetch_fred(series_id, col_rename, start, end, api_key):
    url = "https://api.stlouisfed.org/fred/series/observations"
    params = {
        "series_id": series_id,
        "api_key": api_key,
        "file_type": "json",
        "observation_start": start,
        "observation_end": end,
    }
    try:
        resp = requests.get(url, params=params, timeout=30)
        resp.raise_for_status()
        data = resp.json()
    except Exception as e:
        print(f"  \u26a0\ufe0f  FRED {series_id} error: {e}")
        return pd.DataFrame()

    if "observations" not in data:
        print(f"  \u26a0\ufe0f  FRED {series_id}: {data.get('error_message', 'no observations key')}")
        return pd.DataFrame()

    rows = [
        {"date": o["date"], col_rename: float(o["value"])}
        for o in data["observations"] if o["value"] != "."
    ]
    if not rows:
        return pd.DataFrame()

    return (pd.DataFrame(rows)
              .drop_duplicates("date")
              .sort_values("date")
              .reset_index(drop=True))


# ── 3. BI Rate dari BPS ─────────────────────────────────────────────
def fetch_bi_rate_bps(api_key):
    import requests
    import pandas as pd
    from datetime import datetime

    today = datetime.today()
    year = today.year
    year_bps = year - 1900  # format BPS

    url = (
        f"https://webapi.bps.go.id/v1/api/list/model/data/lang/ind"
        f"/domain/0000/var/379/th/{year_bps}/key/{api_key}"
    )

    try:
        resp = requests.get(url, timeout=30)
        resp.raise_for_status()
        data = resp.json()
    except Exception as e:
        print(f"⚠️ BPS BI Rate fetch error: {e}")
        return pd.DataFrame()

    # Validasi response
    if data.get("status") != "OK":
        print(f"⚠️ BPS status not OK: {data.get('status')}")
        return pd.DataFrame()

    content = data.get("datacontent", {})
    if not content:
        print("⚠️ BPS: datacontent kosong")
        return pd.DataFrame()

    rows = []

    for k, v in content.items():
        try:
            # Extract bulan dari key (robust)
            # handle bulan 1 digit & 2 digit
            last2 = int(k[-2:])
            if 1 <= last2 <= 12:
                month = last2
            else:
                month = int(k[-1])

            # skip "Tahunan" (biasanya 13)
            if month > 12:
                continue

            rows.append({
                "date": f"{year}-{month:02d}-01",
                "bi_rate_pct": float(v)
            })

        except Exception:
            continue

    if not rows:
        print("⚠️ BPS: tidak ada data valid setelah parsing")
        return pd.DataFrame()

    df = (
        pd.DataFrame(rows)
        .drop_duplicates("date")
        .sort_values("date")
        .reset_index(drop=True)
    )

    print(f"✅ BI Rate BPS: {len(df)} rows ({df['date'].min()} → {df['date'].max()})")

    return df


# ── Fetch semua ───────────────────────────────────────────────────────
print("Fetching macro data dari API...")

def _fallback_from_master(col):
    if col in master.columns:
        return (master[["date", col]]
                .dropna(subset=[col])
                .drop_duplicates("date")
                .sort_values("date")
                .reset_index(drop=True))
    return pd.DataFrame()

df_usdidr = fetch_usdidr_bi(_year_start, _yesterday)
if df_usdidr.empty:
    print("  \u26a0\ufe0f  usdidr   : fallback dari master")
    df_usdidr = _fallback_from_master("usdidr")
else:
    print(f"  \u2705 usdidr   : {len(df_usdidr)} rows "
          f"({df_usdidr['date'].min()} \u2192 {df_usdidr['date'].max()})")

df_brent = fetch_fred("DCOILBRENTEU", "brent_usd", _year_start, _yesterday, FRED_API_KEY)
if df_brent.empty:
    print("  \u26a0\ufe0f  brent_usd: fallback dari master")
    df_brent = _fallback_from_master("brent_usd")
else:
    print(f"  \u2705 brent_usd: {len(df_brent)} rows "
          f"({df_brent['date'].min()} \u2192 {df_brent['date'].max()})")

df_coal_m = fetch_fred("PCOALAUUSDM", "coal_usd", _year_start, _yesterday, FRED_API_KEY)
if df_coal_m.empty:
    print("  \u26a0\ufe0f  coal_usd : fallback dari master (monthly)")
    df_coal_m = _fallback_from_master("coal_usd")
else:
    print(f"  \u2705 coal_usd : {len(df_coal_m)} rows bulanan "
          f"({df_coal_m['date'].min()} \u2192 {df_coal_m['date'].max()})")

df_bi_rate = fetch_bi_rate_bps(BPS_API_KEY)
if df_bi_rate.empty:
    print("  \u26a0\ufe0f  bi_rate  : fallback dari master")
    df_bi_rate = _fallback_from_master("bi_rate_pct")
else:
    print(f"  \u2705 bi_rate  : {len(df_bi_rate)} rows "
          f"({df_bi_rate['date'].min()} \u2192 {df_bi_rate['date'].max()})")

# ── Simpan CSV macro ke MACRO_DIR ────────────────────────────────────
Path(MACRO_DIR).mkdir(parents=True, exist_ok=True)
_csv_map = [
    (df_usdidr,  "usdidr.csv",   "usdidr"),
    (df_brent,   "brent.csv",    "brent_usd"),
    (df_coal_m,  "coal.csv",     "coal_usd"),
    (df_bi_rate, "bi_rate.csv",  "bi_rate_pct"),
]
for _df, _fname, _col in _csv_map:
    if not _df.empty and _col in _df.columns:
        _df[["date", _col]].to_csv(f"{MACRO_DIR}/{_fname}", index=False)

# ── Buat macro_ref per tanggal baru ──────────────────────────────────
new_dates_sorted = sorted(new_merged["date"].unique())
macro_ref = pd.DataFrame({"date": new_dates_sorted})

def _ffill_merge_daily(ref_df, src_df, col):
    """Left-join harian lalu forward-fill (untuk data harian/sparse)."""
    if src_df.empty or col not in src_df.columns:
        return ref_df
    ref_df = ref_df.merge(src_df[["date", col]], on="date", how="left")
    ref_df[col] = ref_df[col].ffill().bfill()
    return ref_df

def _ffill_monthly_to_daily(ref_df, src_df, col):
    """Reindex series bulanan ke indeks harian lalu forward-fill."""
    if src_df.empty or col not in src_df.columns:
        return ref_df
    _date_idx = pd.to_datetime(ref_df["date"])
    _s = (src_df
          .assign(date=lambda d: pd.to_datetime(d["date"]))
          .set_index("date")[col]
          .reindex(_date_idx)
          .ffill().bfill())
    ref_df[col] = _s.values
    return ref_df

# Brent: harian — left-join + ffill (bisa ada gap weekend/libur)
macro_ref = _ffill_merge_daily(macro_ref, df_brent,  "brent_usd")

# USDIDR: harian — left-join + ffill (BI tidak publish tiap hari)
macro_ref = _ffill_merge_daily(macro_ref, df_usdidr, "usdidr")

# Coal: bulanan → forward-fill ke harian
macro_ref = _ffill_monthly_to_daily(macro_ref, df_coal_m, "coal_usd")

# BI Rate: bulanan → forward-fill ke harian
macro_ref = _ffill_monthly_to_daily(macro_ref, df_bi_rate, "bi_rate_pct")

# Merge macro ke new_merged
_macro_cols = [c for c in ["usdidr","brent_usd","coal_usd","bi_rate_pct"]
               if c in macro_ref.columns]
new_merged = new_merged.merge(macro_ref[["date"] + _macro_cols], on="date", how="left")

print("\nNull check macro pada baris baru:")
for c in _macro_cols:
    null_pct = new_merged[c].isna().mean() * 100
    icon = "\u2705" if null_pct == 0 else "\u26a0\ufe0f "
    print(f"  {icon} {c:<12}: {null_pct:.1f}% null")

# STEP 7: Merge Sektor dari master
sektor_cols = [c for c in ["ticker","Sektor","SubSektor","Industri","SubIndustri"]
               if c in master.columns]
if len(sektor_cols) > 1:
    sektor_ref = master[sektor_cols].drop_duplicates("ticker").reset_index(drop=True)
    new_merged = new_merged.merge(sektor_ref, on="ticker", how="left")
    print(f"Sektor merged: {new_merged['Sektor'].notna().sum():,} rows punya sektor")

print(f"\nFinal new_merged: {new_merged.shape}")
```
